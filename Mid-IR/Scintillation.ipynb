```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def scintillation_function(x):
    return (1 + 1/x)**(-7/6)

def power_function(x):
    return x**(-7/6)

xs = np.linspace(0.01, 10, 100)
scintillation_values = scintillation_function(xs)
power_values = power_function(xs)

plt.figure(figsize=(10, 5))
plt.plot(xs, scintillation_values, label='Scintillation Function', color='blue')
plt.plot(xs, power_values, label='Power Function', color='orange')
plt.title('Scintillation Function vs Power Function')
plt.legend()
plt.yscale('log')
```
