```python
import Teaching_Siglent_communication as osc
import xy.pyplot as plt
```

```python
scope = osc.connect()
```

```python
scope.clear()  
# osc.set_vertical(inst=scope,channel=1,volts_per_div=5,offset=-14)
# osc.set_vertical(inst=scope,channel=2,volts_per_div=2,offset=-4)
# osc.set_horizontal(inst=scope,seconds_per_div=2e-1,delay=800e-3)
osc.set_trigger(inst=scope, channel=1, level=9.0, slope="RISing")
# Memory depth must be set BEFORE triggering: changing it re-arms the scope,
# which invalidates any record you already captured.
osc.set_memory_depth(inst=scope,points="1M")
osc.single_trigger(inst=scope)
```

```python
data_1 = osc.get_waveform(inst=scope, channel=1)
data_2 = osc.get_waveform(inst=scope, channel=2)
data_3 = osc.get_waveform(inst=scope, channel=3)


time = data_1[0]
v1 = data_1[1]
v2 = data_2[1]
v3 = data_3[1]
```

```python
plt.figure(figsize=[16,9])
plt.plot(time[2000:4000], v1[2000:4000], '.-', label="CH1")
plt.xlabel("time [s]")
plt.ylabel("voltage [V]")
plt.legend()
# plt.ylim([-0.5,1])
```

```python
plt.figure(figsize=[16,9])
plt.plot(time[2000:4000], v2[2000:4000], '.-', label="CH1")
plt.xlabel("time [s]")
plt.ylabel("voltage [V]")
plt.legend()
# plt.ylim([-0.5,1])
```

```python
plt.figure(figsize=[16,9])
plt.plot(time,v3, '.', label="CH3")
plt.xlabel("time [s]")
plt.ylabel("voltage [V]")
plt.legend()
# plt.ylim([-0.5,1])
plt.xlim([0.8,1.2])
```

```python
osc.save_waveform(scope, r".\Data\W1160G1090FF_30_1.csv", time, {1: v1, 2: v2, 3: v3})
```
