```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
S_shot = 1e-15
f_rel_osc_1 = 5e3
f_rel_osc_2 = 6e5

amp_1 = 5e-10
amp_2 = 2e-10

width_1 = 1.5e3
width_2 = 1e5

def rel_osc_amp (f,f_rel_osc,amp,width):

    return amp/(1 + ((f-f_rel_osc)/width)**2)

f = np.logspace(3,7,int(1e4))

S_RIN = []

for freq in f:
    S_RIN.append ( S_shot + rel_osc_amp(freq,f_rel_osc_1,amp_1,width_1) + rel_osc_amp(freq,f_rel_osc_2,amp_2,width_2) )

S_RIN_dB = 10*np.log10(S_RIN)
```

```python
plt.figure()
plt.plot(f,S_RIN_dB)
plt.xscale('log')
plt.ylim(-160,-80)
```
