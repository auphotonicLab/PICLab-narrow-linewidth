```python
import numpy as np
import xy.pyplot as plt
import h5py
```

```python
def dB_to_linear(dB):
    """Convert dB to linear scale."""
    return 10 ** (dB / 10)

def linear_to_dB(linear):
    """Convert linear scale to dB."""
    return 10 * np.log10(linear)
```

```python
#RIN comparison

folder_RIN_GUI_n1 = r"C:\Users\au617810\OneDrive - Aarhus universitet\O-drive - Jeppe\HF Python control test\NKT 10mW EDFA 1090mA\GUI RIN n=1"
folder_RIN_GUI_n20 = r"C:\Users\au617810\OneDrive - Aarhus universitet\O-drive - Jeppe\HF Python control test\NKT 10mW EDFA 1090mA\GUI RIN n=20"


files_n1 = []
files_n20 = []

n1_data = []
n20_data = []

for i in range(20):
    files_n1.append(folder_RIN_GUI_n1 + fr'\PSD_{i}.txt')

    n1_data.append(np.loadtxt(files_n1[i],skiprows=13,delimiter=','))
    
for i in range(3):
    files_n20.append(folder_RIN_GUI_n20 + fr'\PSD_{i}.txt')
    
    n20_data.append(np.loadtxt(files_n20[i],skiprows=13,delimiter=','))


files_n1_python = r"C:\Users\au617810\OneDrive - Aarhus universitet\O-drive - Jeppe\HF Python control test\NKT 10mW EDFA 1090mA\RIN Python n=20\2026-10-21_18-02-20_avg20.h5"

files_n20_python = r"C:\Users\au617810\OneDrive - Aarhus universitet\O-drive - Jeppe\HF Python control test\NKT 10mW EDFA 1090mA\RIN Python n=20\2026-10-21_18-02-20_avg20_RIN.txt"




n1_data_python = h5py.File(files_n1_python)['noise/captures']


n20_data_python = np.loadtxt(files_n20_python,skiprows=13,delimiter=',')

```

```python
#Extracting GUI data. Averaging n1 data both as dB and linear scale.
x_n1_data = []


y_n1_data = []
y_linear_n1_data = []

for i in range(20):
    
    x_n1_data.append(n1_data[i][:,0])

    y_n1_data.append(n1_data[i][:,1])
    y_linear_n1_data.append(dB_to_linear(n1_data[i][:,1])) #Since this is in dB/Hz

avg_y_n1_data = np.mean(y_n1_data, axis=0)
avg_y_linear_n1_data= linear_to_dB(np.mean(y_linear_n1_data, axis=0))
```

```python
#Extracting Python data. Averaging n1 data both as dB and linear scale.
n1_data_python_arr = n1_data_python[:]

n1_data_linear_python_arr = dB_to_linear(n1_data_python[:])

n1_data_python_avg = np.mean(n1_data_python_arr,axis=0)

n1_data_linear_python_avg = linear_to_dB(np.mean(n1_data_linear_python_arr,axis=0))
```

```python
#Plotting Python n20 and n1 data (averaged both dB and linear scale) for comparison.
plt.figure()
plt.plot(n20_data_python[:,0], n20_data_python[:,1], label='Python n20 dB/Hz avg', color='red', alpha=0.5)

plt.plot(n20_data_python[:,0], n1_data_python_avg, label='Python n1 dB/Hz avg', color='blue',alpha=0.5,linestyle='--')
plt.plot(n20_data_python[:,0], n1_data_linear_python_avg, label='Python n1 linear avg', color='green',alpha=0.5,linestyle='--')
# plt.yscale('log')
plt.xscale('log')
plt.xlabel('Frequency (Hz)')
plt.ylabel('dB/Hz')
plt.title('Python n20 vs n1 averaged dB/Hz and linear')
```

```python
#GUI averaging comparison

plt.figure()
plt.plot(x_n1_data[0],avg_y_n1_data, label='dB avg')
plt.plot(x_n1_data[0],avg_y_linear_n1_data, label = 'linear squared avg ')

plt.plot(n20_data[0][:,0],n20_data[0][:,1], label = 'GUI n20',color='green', alpha=0.5, linestyle='--')


plt.ylabel('dB/Hz')
plt.xlabel('Hz')
plt.xscale('log')
plt.title('GUI averaging comparison')
```

```python
plt.figure()
plt.plot(x_n1_data[0],avg_y_n1_data-avg_y_linear_n1_data, label='dB/Hz avg - linear avg')

plt.plot(n20_data[0][:,0],n20_data[0][:,1]-avg_y_linear_n1_data, label = 'GUI n20 - linear avg')

print('Subtracting then mean:')
print('Amp avg squared - power avg',np.mean(avg_y_n1_data-avg_y_linear_n1_data)), 'dB'
print('GUI n20 - linear avg',np.mean(n20_data[0][:,1]-avg_y_linear_n1_data), 'dB')

plt.ylabel('dB')
plt.xlabel('Hz')
# plt.yscale('log')
# plt.xscale('log')
plt.title('GUI averaging comparison, subtracting power avg')
plt.ylim([-10,10])
```

```python
#Python vs GUI comparison


plt.figure()
plt.plot(x_n1_data[0],avg_y_n1_data, label='GUI n1 dB avg')
plt.plot(x_n1_data[0],avg_y_linear_n1_data, label = 'GUI n1 linear avg ')

plt.plot(n20_data[0][:,0],n20_data[0][:,1], label = 'GUI n20 avg',color='blue', alpha=0.5, linestyle='--')

plt.plot(n20_data_python[:,0], n20_data_python[:,1], label='Python n20 avg', color='green', alpha=0.5, linestyle='--')


plt.ylabel('dB/Hz')
plt.xlabel('Hz')
plt.xscale('log')
plt.title('Python vs GUI averaging comparison')
```

```python
#Python vs GUI subtracting Python n20 Hz2/Hz from all data sets

plt.figure()
plt.plot(x_n1_data[0],avg_y_n1_data - n20_data_python[:,1], label='dB avg')
plt.plot(x_n1_data[0],avg_y_linear_n1_data - n20_data_python[:,1], label = 'linear avg ')

plt.plot(n20_data[0][:,0],n20_data[0][:,1] - n20_data_python[:,1], label = 'GUI n20')

print('Subtracting then mean:')
print('dB avg - Python n20',np.mean(avg_y_n1_data - n20_data_python[:,1]))
print('linear avg - Python n20',np.mean(avg_y_linear_n1_data - n20_data_python[:,1]))
print('GUI n20 - Python n20',np.mean(n20_data[0][:,1] - n20_data_python[:,1]))


plt.ylabel('Hz2/Hz')
plt.xlabel('Hz')
# plt.yscale('log')
# plt.xscale('log')
plt.title('Subtracting Python n20 from all data sets')
plt.ylim([-25,25])
```
