```python
import numpy as np
import xy.pyplot as plt
import h5py
```

```python
def linear_to_dB(linear):
    """Convert linear scale to dB."""
    return 10 * np.log10(linear)
```

```python
#FN comparison

folder_FN_GUI_n1 = r"C:\Users\au617810\OneDrive - Aarhus universitet\O-drive - Jeppe\HF Python control test\NKT 10mW EDFA 1090mA\GUI FN n = 1"
folder_FN_GUI_n20 = r"C:\Users\au617810\OneDrive - Aarhus universitet\O-drive - Jeppe\HF Python control test\NKT 10mW EDFA 1090mA\GUI FN n=20"


files_n1 = []
files_n1_Hz2 = []
files_n20 = []
files_n20_Hz2 = []

n1_data = []
n1_Hz2_data = []
n20_data = []
n20_Hz2_data = []

for i in range(20):
    files_n1.append(folder_FN_GUI_n1 + fr'\PSD_{i}.txt')
    files_n1_Hz2.append(folder_FN_GUI_n1 + fr'\Hz2_PSD_{i}.txt')

    n1_data.append(np.loadtxt(files_n1[i],skiprows=13,delimiter=','))
    n1_Hz2_data.append(np.loadtxt(files_n1_Hz2[i],skiprows=13,delimiter=','))
    
for i in range(3):
    files_n20.append(folder_FN_GUI_n20 + fr'\PSD_{i}.txt')
    files_n20_Hz2.append(folder_FN_GUI_n20 + fr'\Hz2_PSD_{i}.txt')
    
    n20_data.append(np.loadtxt(files_n20[i],skiprows=13,delimiter=','))
    n20_Hz2_data.append(np.loadtxt(files_n20_Hz2[i],skiprows=13,delimiter=','))


files_n1_python = r"C:\Users\au617810\OneDrive - Aarhus universitet\O-drive - Jeppe\HF Python control test\NKT 10mW EDFA 1090mA\FN Python n=20\2026-10-21_16-39-20_avg20.h5"
files_n1_Hz2_python = r"C:\Users\au617810\OneDrive - Aarhus universitet\O-drive - Jeppe\HF Python control test\NKT 10mW EDFA 1090mA\FN Hz2 Python n=20\2026-10-21_16-42-02_avg20.h5"

files_n20_python = r"C:\Users\au617810\OneDrive - Aarhus universitet\O-drive - Jeppe\HF Python control test\NKT 10mW EDFA 1090mA\FN Python n=20\2026-10-21_16-39-20_avg20_PSD.txt"
files_n20_Hz2_python = r"C:\Users\au617810\OneDrive - Aarhus universitet\O-drive - Jeppe\HF Python control test\NKT 10mW EDFA 1090mA\FN Hz2 Python n=20\2026-10-21_16-42-02_avg20_PSD.txt"


n1_data_python = h5py.File(files_n1_python)['noise/captures']
n1_Hz2_data_python = h5py.File(files_n1_Hz2_python)['noise/captures']


n20_data_python = np.loadtxt(files_n20_python,skiprows=13,delimiter=',')
n20_Hz2_data_python = np.loadtxt(files_n20_Hz2_python,skiprows=13,delimiter=',')

```

```python
#Extracting GUI data. Then averaging n1 in both amp and power.

x_n1_data = []
x_n1_Hz2_data = []

y_n1_data = []
y_squared_n1_data = []
y_n1_Hz2_data = []
for i in range(20):
    
    x_n1_data.append(n1_data[i][:,0])
    x_n1_Hz2_data.append(n1_Hz2_data[i][:,0])

    y_n1_data.append(n1_data[i][:,1])
    y_squared_n1_data.append(n1_data[i][:,1]**2) #Since this is in Hz/sq(Hz)
    y_n1_Hz2_data.append(n1_Hz2_data[i][:,1])

```

```python
#Extracting Python data. Then averaging n1 in power.

n1_data_python_arr = n1_data_python[:]

n1_data_python_avg = np.mean(n1_data_python_arr,axis=0)

n1_Hz2_data_python_arr = n1_Hz2_data_python[:]

n1_Hz2_data_python_avg = np.mean(n1_Hz2_data_python_arr,axis=0)
```

```python
#Check that Python data gives the same results even if GUI is run in Hz/sq(Hz) or Hz2/Hz.
#Also check that averaging is done in Hz^2/Hz (power) and not in Hz/sq(Hz) (amplitude).
plt.figure()
plt.plot(n20_data_python[:,0], n20_data_python[:,1], label='Python n20 Hz/sq(Hz) avg', color='green', alpha=0.5)
plt.plot(n20_Hz2_data_python[:,0], n20_Hz2_data_python[:,1], label='Python n20 Hz2/Hz avg', color='black', alpha=0.5)

plt.plot(n20_Hz2_data_python[:,0], n1_data_python_avg, label='Python n1 Hz/sq(Hz) avg', color='blue',alpha=0.5)
plt.plot(n20_Hz2_data_python[:,0], n1_Hz2_data_python_avg, label='Python n1 Hz2/Hz avg', color='red',alpha=0.5)
plt.yscale('log')
plt.xscale('log')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Hz2/Hz)')
plt.title('Python n1 vs n20 (Hz2/Hz and Hz/sq(Hz) shown on GUI meanwhile)')
```

```python
#Checking GUI data run in Hz/sq(Hz) or Hz2/Hz. Also plotting the square of the Hz/sq(Hz) data.
#As expected, the squared Hz/sq(Hz) data is equivalent to the Hz2/Hz data.
for i in range(20):
    plt.plot(x_n1_data[i],y_n1_data[i],'b--',label='GUI Hz/sq(Hz) n=1' if i==0 else "")
    # plt.plot(x_n1_data[i],y_squared_n1_data[i],'--')
    plt.plot(x_n1_data[i],y_squared_n1_data[i],'g',alpha=0.5, label='GUI Hz/sq(Hz) squared n=1' if i==0 else "")
    plt.plot(x_n1_data[i],y_n1_Hz2_data[i],'r', alpha=0.5, label='GUI Hz2/Hz n=1' if i==0 else "")

plt.xscale('log')
plt.yscale('log')
plt.ylim([1e-5,1e6])
plt.xlim([5e1,1e7])
plt.xlabel('Frequency (Hz)')
plt.ylabel('Hz2/Hz and Hz/sq(Hz)')
plt.title('GUI n=1 Hz2/Hz, Hz/sq(Hz), and Hz/sq(Hz) squared comparison')
```

```python
#Comparing the GUI averaging methods.
#Compare n1 avg of amplitude, then square it, vs avg of amplitude squared, vs avg of power (Hz2/Hz)
#Then compare that to GUI n20 data for both Hz/sq(Hz) and Hz2/Hz.

avg_y_n1_data = (np.mean(y_n1_data,axis=0) )**2 #To get it in Hz^2/Hz
avg_y_squared_n1_data = np.mean(y_squared_n1_data,axis=0)
avg_y_n1_Hz2_data = np.mean(y_n1_Hz2_data,axis=0)

plt.figure()
plt.plot(x_n1_data[0],avg_y_n1_data, label='Amp avg squared')
plt.plot(x_n1_data[0],avg_y_squared_n1_data, label = 'Amp squared avg ')
plt.plot(x_n1_data[0],avg_y_n1_Hz2_data, label = 'power avg ')

plt.plot(n20_Hz2_data[0][:,0],n20_Hz2_data[0][:,1], label = 'GUI n20 Hz2/Hz')
plt.plot(n20_data[0][:,0],(n20_data[0][:,1])**2, label = 'GUI n20 Hz/sq(Hz) squared')


plt.ylabel('Hz2/Hz')
plt.xlabel('Hz')
plt.yscale('log')
plt.xscale('log')
plt.title('GUI averaging comparison')
```

```python
#Comparing the GUI averaging methods as above. Here we subtract the power avg from all data sets to see how they compare.
#Also converting the avg data to dB and then subtracting once more.

#Linear, Hz2/Hz scale
plt.figure()
plt.plot(x_n1_data[0],avg_y_n1_data-avg_y_n1_Hz2_data, label='Amp avg squared - power avg')
plt.plot(x_n1_data[0],avg_y_squared_n1_data-avg_y_n1_Hz2_data, label = 'Amp squared avg - power avg')

plt.plot(n20_Hz2_data[0][:,0],n20_Hz2_data[0][:,1]-avg_y_n1_Hz2_data, label = 'GUI n20 Hz2/Hz - power avg')
plt.plot(n20_data[0][:,0],(n20_data[0][:,1])**2-avg_y_n1_Hz2_data, label = 'GUI n20 Hz/sq(Hz) squared - power avg')

print('Subtracting then mean:')
print('Amp avg squared - power avg',np.mean(avg_y_n1_data-avg_y_n1_Hz2_data))
print('Amp squared avg - power avg',np.mean(avg_y_squared_n1_data-avg_y_n1_Hz2_data))
print('GUI n20 Hz2/Hz - power avg',np.mean(n20_Hz2_data[0][:,1]-avg_y_n1_Hz2_data))
print('GUI n20 Hz/sq(Hz) squared - power avg',np.mean((n20_data[0][:,1])**2-avg_y_n1_Hz2_data))

plt.ylabel('Hz2/Hz')
plt.xlabel('Hz')
# plt.yscale('log')
# plt.xscale('log')
plt.title('GUI averaging comparison, subtracting power avg')
plt.ylim([-25,25])

#Convert avg data to dB, then subtracting the power avg in dB from all data sets.

plt.figure()
plt.plot(x_n1_data[0],linear_to_dB(avg_y_n1_data)-linear_to_dB(avg_y_n1_Hz2_data), label='Amp avg squared - power avg')
plt.plot(x_n1_data[0],linear_to_dB(avg_y_squared_n1_data)-linear_to_dB(avg_y_n1_Hz2_data), label = 'Amp squared avg - power avg')

plt.plot(n20_Hz2_data[0][:,0],linear_to_dB(n20_Hz2_data[0][:,1])-linear_to_dB(avg_y_n1_Hz2_data), label = 'GUI n20 Hz2/Hz - power avg')
plt.plot(n20_data[0][:,0],linear_to_dB((n20_data[0][:,1])**2)-linear_to_dB(avg_y_n1_Hz2_data), label = 'GUI n20 Hz/sq(Hz) squared - power avg')

print('\nSubtracting in dB then mean:')
print('Amp avg squared - power avg',np.mean(linear_to_dB(avg_y_n1_data)-linear_to_dB(avg_y_n1_Hz2_data)))
print('Amp squared avg - power avg',np.mean(linear_to_dB(avg_y_squared_n1_data)-linear_to_dB(avg_y_n1_Hz2_data)))
print('GUI n20 Hz2/Hz - power avg',np.mean(linear_to_dB(n20_Hz2_data[0][:,1])-linear_to_dB(avg_y_n1_Hz2_data)))
print('GUI n20 Hz/sq(Hz) squared - power avg',np.mean(linear_to_dB((n20_data[0][:,1])**2)-linear_to_dB(avg_y_n1_Hz2_data)))

plt.ylabel('dB')
plt.xlabel('Hz')
# plt.yscale('log')
# plt.xscale('log')
plt.title('GUI averaging comparison, subtracting power avg')
plt.ylim([-8,8])

```

```python
#Comparing GUI n20 averaging method.
#Check Hz/sq(Hz) vs Hz^2/Hz n20 data directly.
#Also converting to dB and then subtracting.

#Linear scale Hz2/Hz
plt.figure()
plt.plot(n20_Hz2_data[0][:,0],(n20_data[0][:,1])**2 - n20_Hz2_data[0][:,1], label = 'Difference GUI n20 Hz/sq(Hz) squared - n20 Hz^2/Hz')
plt.ylim([-25,25])
print('Mean difference GUI n20 Hz/sq(Hz) squared - n20 Hz^2/Hz',np.mean((n20_data[0][:,1])**2- n20_Hz2_data[0][:,1]))


#dB scale
plt.figure()
plt.plot(n20_Hz2_data[0][:,0],linear_to_dB((n20_data[0][:,1])**2) - linear_to_dB(n20_Hz2_data[0][:,1]), label = 'Difference GUI n20 Hz/sq(Hz) squared - n20 Hz^2/Hz')
plt.ylim([-8,8])
print('\nConvert to dB then subtract:')
print('Mean difference GUI n20 Hz/sq(Hz) squared - n20 Hz^2/Hz',np.mean(linear_to_dB((n20_data[0][:,1])**2)- linear_to_dB(n20_Hz2_data[0][:,1])))
```

```python
#Python vs GUI comparison
#Plotting the GUI n1 averaging methods vs the Python n1 averaging methods,
#Also comparing to GUI n20 data for both Hz/sq(Hz) and Hz2/Hz.

plt.figure()
plt.plot(x_n1_data[0],avg_y_n1_data, label='Amp avg squared')
plt.plot(x_n1_data[0],avg_y_squared_n1_data, label = 'Amp squared avg ')
plt.plot(x_n1_data[0],avg_y_n1_Hz2_data, label = 'power avg ')

plt.plot(n20_Hz2_data[0][:,0],n20_Hz2_data[0][:,1], label = 'GUI n20 Hz2/Hz')
plt.plot(n20_data[0][:,0],(n20_data[0][:,1])**2, label = 'GUI n20 Hz/sq(Hz) squared')

plt.plot(n20_data_python[:,0], n20_data_python[:,1], label='Python n20 Hz/sq(Hz) avg', color='green', alpha=0.5)
plt.plot(n20_Hz2_data_python[:,0], n20_Hz2_data_python[:,1], label='Python n20 Hz2/Hz avg', color='black', alpha=0.5)


plt.ylabel('Hz2/Hz')
plt.xlabel('Hz')
plt.yscale('log')
plt.xscale('log')
plt.title('Python vs GUI averaging comparison')
```

```python
#Python vs GUI subtracting Python n20 Hz2/Hz from all data sets.
#Since we know that Python gives power avg no matter the GUI setting,
#we now check that which GUI averaging method is closest to the Python n20 Hz2/Hz data.
#Also converting the averages to dB scale and then subtracting. 

#Result is that the GUI amplitude averages no matter the GUI setting.
#From the data it is clear that squaring the amplitude of the Hz/sq(Hz)
#data gives the same result as the Hz2/Hz data,
#and averaging in those units are equivalent to the Python n20 Hz2/Hz data.



#Linear scale Hz2/Hz
plt.figure()
plt.plot(x_n1_data[0],avg_y_n1_data - n20_Hz2_data_python[:,1], label='Amp avg squared')
plt.plot(x_n1_data[0],avg_y_squared_n1_data - n20_Hz2_data_python[:,1], label = 'Amp squared avg ')
plt.plot(x_n1_data[0],avg_y_n1_Hz2_data - n20_Hz2_data_python[:,1], label = 'power avg ')

plt.plot(n20_Hz2_data[0][:,0],n20_Hz2_data[0][:,1] - n20_Hz2_data_python[:,1], label = 'GUI n20 Hz2/Hz')
plt.plot(n20_data[0][:,0],(n20_data[0][:,1])**2 - n20_Hz2_data_python[:,1], label = 'GUI n20 Hz/sq(Hz) squared')

plt.plot(n20_data_python[:,0], n20_data_python[:,1] - n20_Hz2_data_python[:,1], label='Python n20 Hz/sq(Hz) avg', color='green', alpha=0.5)

print('Subtracting then mean:')
print('Amp avg squared - Python n20 Hz2/Hz',np.mean(avg_y_n1_data - n20_Hz2_data_python[:,1]))
print('Amp squared avg - Python n20 Hz2/Hz',np.mean(avg_y_squared_n1_data - n20_Hz2_data_python[:,1]))
print('power avg - Python n20 Hz2/Hz',np.mean(avg_y_n1_Hz2_data - n20_Hz2_data_python[:,1]))
print('GUI n20 Hz2/Hz - Python n20 Hz2/Hz',np.mean(n20_Hz2_data[0][:,1] - n20_Hz2_data_python[:,1]))
print('GUI n20 Hz/sq(Hz) squared - Python n20 Hz2/Hz',np.mean((n20_data[0][:,1])**2 - n20_Hz2_data_python[:,1]))
print('Python n20 Hz/sq(Hz) avg - Python n20 Hz2/Hz',np.mean(n20_data_python[:,1] - n20_Hz2_data_python[:,1]))


plt.ylabel('Hz2/Hz')
plt.xlabel('Hz')
# plt.yscale('log')
# plt.xscale('log')
plt.title('Subtracting Python n20 Hz2/Hz from all data sets')
plt.ylim([-25,25])


#dB scale

plt.figure()
plt.plot(x_n1_data[0],linear_to_dB(avg_y_n1_data) - linear_to_dB(n20_Hz2_data_python[:,1]), label='Amp avg squared')
plt.plot(x_n1_data[0],linear_to_dB(avg_y_squared_n1_data) - linear_to_dB(n20_Hz2_data_python[:,1]), label = 'Amp squared avg ')
plt.plot(x_n1_data[0],linear_to_dB(avg_y_n1_Hz2_data) - linear_to_dB(n20_Hz2_data_python[:,1]), label = 'power avg ')

plt.plot(n20_Hz2_data[0][:,0],linear_to_dB(n20_Hz2_data[0][:,1]) - linear_to_dB(n20_Hz2_data_python[:,1]), label = 'GUI n20 Hz2/Hz')
plt.plot(n20_data[0][:,0],linear_to_dB((n20_data[0][:,1])**2) - linear_to_dB(n20_Hz2_data_python[:,1]), label = 'GUI n20 Hz/sq(Hz) squared')

plt.plot(n20_data_python[:,0], linear_to_dB(n20_data_python[:,1]) - linear_to_dB(n20_Hz2_data_python[:,1]), label='Python n20 Hz/sq(Hz) avg', color='green', alpha=0.5)

print('\nConvert to dB and subtracting then mean:')
print('Amp avg squared - Python n20 Hz2/Hz',np.mean(linear_to_dB(avg_y_n1_data) - linear_to_dB(n20_Hz2_data_python[:,1])))
print('Amp squared avg - Python n20 Hz2/Hz',np.mean(linear_to_dB(avg_y_squared_n1_data) - linear_to_dB(n20_Hz2_data_python[:,1])))
print('power avg - Python n20 Hz2/Hz',np.mean(linear_to_dB(avg_y_n1_Hz2_data) - linear_to_dB(n20_Hz2_data_python[:,1])))
print('GUI n20 Hz2/Hz - Python n20 Hz2/Hz',np.mean(linear_to_dB(n20_Hz2_data[0][:,1]) - linear_to_dB(n20_Hz2_data_python[:,1])))
print('GUI n20 Hz/sq(Hz) squared - Python n20 Hz2/Hz',np.mean(linear_to_dB((n20_data[0][:,1])**2) - linear_to_dB(n20_Hz2_data_python[:,1])))
print('Python n20 Hz/sq(Hz) avg - Python n20 Hz2/Hz',np.mean(linear_to_dB(n20_data_python[:,1]) - linear_to_dB(n20_Hz2_data_python[:,1])))


plt.ylabel('dB')
plt.xlabel('Hz')
# plt.yscale('log')
# plt.xscale('log')
plt.title('Subtracting Python n20 Hz2/Hz from all data sets')
plt.ylim([-8,8])
```
